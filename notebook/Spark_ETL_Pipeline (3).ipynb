```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/workspace/stream/streaming/Online Retail.csv")
display(df)
```

```python
# Data Cleaning
from pyspark.sql.functions import trim, col, to_timestamp, coalesce

# Drop nulls & bad quantities
df = df.dropna()
df = df.filter(df.Quantity > 0)

# Fix date
df = df.withColumn("InvoiceDate", trim("InvoiceDate"))
df = df.withColumn("InvoiceDate",
    coalesce(
        to_timestamp("InvoiceDate", "MM/dd/yyyy HH:mm:ss"),
        to_timestamp("InvoiceDate", "dd/MM/yyyy HH:mm:ss")
    )
)

# Add Revenue
df = df.withColumn("Revenue", col("Quantity") * col("UnitPrice"))
print("Cleaned row count:", df.count())
display(df)
```

```python
from pyspark.sql.functions import sum as spark_sum

sales_by_country = df.groupBy("Country") \
    .agg(spark_sum("Revenue").alias("TotalRevenue")) \
    .orderBy("TotalRevenue", ascending=False) \
    .limit(10)

display(sales_by_country)
# Chart settings → Bar Chart | Keys: Country | Values: TotalRevenue
```

```python
from pyspark.sql.functions import month

df = df.withColumn("Month", month("InvoiceDate"))

monthly_sales = df.groupBy("Month") \
    .agg(spark_sum("Revenue").alias("MonthlyRevenue")) \
    .orderBy("Month")

display(monthly_sales)
# Chart settings → Line Chart | Keys: Month | Values: MonthlyRevenue
```

```python
top_products = df.groupBy("Description") \
    .agg(spark_sum("Quantity").alias("TotalSold")) \
    .orderBy("TotalSold", ascending=False) \
    .limit(10)

display(top_products)
# Chart settings → Bar Chart | Keys: Description | Values: TotalSold
```

```python
from pyspark.sql.functions import avg, round as spark_round

avg_order = df.groupBy("Country") \
    .agg(spark_round(avg("Revenue"), 2).alias("AvgOrderValue")) \
    .orderBy("AvgOrderValue", ascending=False) \
    .limit(10)

display(avg_order)
# Chart settings → Bar Chart | Keys: Country | Values: AvgOrderValue
```

```python
top_customers = df.groupBy("CustomerID") \
    .agg(spark_sum("Revenue").alias("TotalSpent")) \
    .orderBy("TotalSpent", ascending=False) \
    .limit(10)

display(top_customers)
# Bar Chart | Keys: CustomerID | Values: TotalSpent
```

```python
from pyspark.sql.functions import dayofweek, count

df = df.withColumn("DayOfWeek", dayofweek("InvoiceDate"))

day_orders = df.groupBy("DayOfWeek") \
    .agg(count("InvoiceNo").alias("TotalOrders")) \
    .orderBy("DayOfWeek")

display(day_orders)
# Line Chart | Keys: DayOfWeek | Values: TotalOrders
# 1=Sun, 2=Mon ... 7=Sat
```

```python
from pyspark.sql.functions import hour

df = df.withColumn("Hour", hour("InvoiceDate"))

hourly_sales = df.groupBy("Hour") \
    .agg(spark_sum("Revenue").alias("HourlyRevenue")) \
    .orderBy("Hour")

display(hourly_sales)
# Line Chart | Keys: Hour | Values: HourlyRevenue
# Shows which hours of the day drive the most sales
```

```python
from pyspark.sql.functions import sum as spark_sum, countDistinct, month

monthly_orders = df.groupBy("Month") \
    .agg(
        spark_sum("Revenue").cast("double").alias("MonthlyRevenue"),
        countDistinct("InvoiceNo").alias("TotalOrders"),
        countDistinct("CustomerID").alias("UniqueCustomers")
    ) \
    .orderBy("Month") \
    .na.drop()  # removes any null Month rows

display(monthly_orders)
```

```python
df.write \
    .mode("overwrite") \
    .partitionBy("Month") \
    .parquet("/Volumes/workspace/stream/streaming/clean_retail_data")

print(" Data saved successfully!")

df_saved = spark.read.parquet("/Volumes/workspace/stream/streaming/clean_retail_data")
print("Saved row count:", df_saved.count())
display(df_saved)
```

```python
from pyspark.sql.functions import sum as spark_sum, countDistinct, count, avg, round as spark_round, month, hour, dayofweek, when, col

#  KPIs 
total_revenue     = df.agg(spark_sum("Revenue")).collect()[0][0]
total_orders      = df.select(countDistinct("InvoiceNo")).collect()[0][0]
total_customers   = df.select(countDistinct("CustomerID")).collect()[0][0]
total_products    = df.select(countDistinct("Description")).collect()[0][0]
avg_order_value   = round(total_revenue / total_orders, 2)

print("=" * 50)
print("       ETL PIPELINE DASHBOARD SUMMARY")
print("=" * 50)
print(f"  Total Revenue       : £{total_revenue:,.2f}")
print(f"  Total Orders        : {total_orders:,}")
print(f"  Unique Customers    : {total_customers:,}")
print(f"  Unique Products     : {total_products:,}")
print(f"  Avg Order Value     : £{avg_order_value:,.2f}")
print("=" * 50)

# ── Top 5 Countries 
print("\n Top 5 Countries by Revenue:")
df.groupBy("Country") \
  .agg(spark_sum("Revenue").alias("TotalRevenue")) \
  .orderBy("TotalRevenue", ascending=False) \
  .limit(5) \
  .show(truncate=False)

# ── Best Month 
best_month = df.groupBy("Month") \
    .agg(spark_sum("Revenue").alias("MonthlyRevenue")) \
    .orderBy("MonthlyRevenue", ascending=False) \
    .limit(1) \
    .collect()[0]

print(f" Best Sales Month     : Month {best_month['Month']} — £{best_month['MonthlyRevenue']:,.2f}")

# ── Best Hour
df = df.withColumn("Hour", hour("InvoiceDate"))
best_hour = df.groupBy("Hour") \
    .agg(spark_sum("Revenue").alias("HourlyRevenue")) \
    .orderBy("HourlyRevenue", ascending=False) \
    .limit(1) \
    .collect()[0]

print(f"Peak Sales Hour      : {best_hour['Hour']}:00 — £{best_hour['HourlyRevenue']:,.2f}")

# ── Top Product ─
top_product = df.groupBy("Description") \
    .agg(spark_sum("Quantity").alias("TotalSold")) \
    .orderBy("TotalSold", ascending=False) \
    .limit(1) \
    .collect()[0]

print(f" Best Selling Product : {top_product['Description']} ({top_product['TotalSold']:,} units)")

# ── Weekday vs Weekend 
df = df.withColumn("DayType",
    when(dayofweek("InvoiceDate").isin([1,7]), "Weekend").otherwise("Weekday"))

print("\n Weekday vs Weekend Revenue:")
df.groupBy("DayType") \
  .agg(spark_sum("Revenue").alias("TotalRevenue")) \
  .show(truncate=False)

# ── Repeat vs One-Time Customers 
customer_orders = df.groupBy("CustomerID") \
    .agg(countDistinct("InvoiceNo").alias("OrderCount"))

customer_summary = customer_orders.withColumn(
    "CustomerType",
    when(col("OrderCount") > 1, "Repeat").otherwise("One-Time")
).groupBy("CustomerType").agg(count("CustomerID").alias("Count"))

print("\n Customer Loyalty:")
customer_summary.show(truncate=False)

print("=" * 50)
print("Dashboard Complete!")
print("=" * 50)
```
